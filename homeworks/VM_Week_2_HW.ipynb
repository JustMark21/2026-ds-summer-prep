# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
with open("../Week-02/intermediate/data/lil-wayne.txt", "r", encoding="utf-8") as readit:
    chars = {}
    text = readit.read().upper()
    for i in text:
        if i != " ":
            chars[i] = chars.get(i, 0) + 1

unique_chars = set(chars.keys())
max_char = ""
max_count = 0

for char in unique_chars:
    current_count = chars[char]
    if current_count > max_count:
        max_count = current_count
        max_char = char

print(f"Char: '{max_char}', with '{max_count}' times appear in array")
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]


````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]

def dev_by_three(mylist):
    new_list = []
    for i in mylist:
        if i % 3 == 0:
            new_list.append(i)
    return new_list

print(dev_by_three(original_list))
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.

```

list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400

```

##  #11 Check if the value 999 exists in the dicitionary below.  If so, print true, if not print false.

Input: sample_dict = {'a': 100, 'b': 999, 'c': 300}

Expected Output:  True

````python
sample_dict = {'a': 100, 'b': 999, 'c': 300}

class Find_it():

    def is_match(self, my_num, my_dict):
        for val in my_dict.values():
            if val == my_num:
                return True
        return False

my_var = Find_it()

bool = my_var.is_match(999, sample_dict)
if bool == False:
    print("false")
else:
    print("true")
````
